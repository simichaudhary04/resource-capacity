#Resource capacity planning and system tunning for high performance system:

1. How can you identify CPU and I/O bottlenecks in a Linux server
using uptime and load averages?
  -Uptime shows load average of last 1, 5, 15 mins. load average is the number of processes waiting for CPU.
  
   ▶CPU Bottleneck: If load is more than your cores, system is overloaded. Check top command. If %us and %sy is high, it means CPU is doing too much work. This is CPU bottleneck.

 2. I/O Bottleneck:
If load is very high but CPU is free, check top command. If %wa is high, it means CPU is waiting for disk. This is I/O bottleneck.

2. How do you identify memory issues such as high swap usage
and low free memory in Linux?
   - o check memory issues in Linux, we use the free -h command. It shows total, used, free, and swap memory.

  1. Low Free Memory:
  If free and available memory is very low  it means system has low memory. You can find which process is using more memory with top command, press Shift + M to sort by memory.

  2. High Swap Usage:
Swap is used when RAM is full.  if swap used is high and it keeps increasing, it is a problem. You can check this with
 vmstat

 1. If si (swap in) and so (swap out) values are continuously high, it means system is constantly swapping, which makes server slow.

3.What is the purpose of sar and Prometheus in system
monitoring?

 -SAR - Sar is a command line that collects data from real time or historical logs like CPU, memory, disk I/O. So that it can help in monitoring the system and identify performance bottlenecks. (Ex. sar -u 5)

 Prometheus - Prometheus is a monitoring tool that collects time series data and allows you to monitor your system over weeks, months or even years.Its typically paired with Grafana for data visualizationBenefits - Prometheus can also trigger alerts when preferred conditions like (resource exhaustion) are met.

4. How can CPU usage be optimized in Linux systems?
  - Optimizing CPU Usage - is about ensuring that the system's CPU is being used efficiently. When the CPU is overloaded performance degrades.

   We can utilize two advanced mechanisms for it:CPU Affinity and Task Offloading.

  1. CPU Affinity: it also known as CPU pinning.this can optimize performance as the core become more familiar,with the process data.

  purpose of it:
  1. Performance optimization.
  2. Preventing overloads

  command for it: taskset -cp 0,1 <PID>

  2.Task Offloading: move a task to another resource like gpu or cloud to offload the processing.

  purpose of it:

  1.Freeing resources.

  2. performance boost.

  3.scalability.

  command for it: scp <source> <Destination>.

5.  How do you tune Linux kernel parameters using sysctl?
   - systemctl is used to view and change linux kernel parameters at runtime.

   Making Changes to Kernel -
   
   View Parameter: sysctl -a
   
   Modify Parameter: sysctl -w
   
   Verify Change: sysctl <parameter
   
   >Make Persistent: Edit /etc/sysctl.conf
   
   Apply Permanently: sudo sysctl -p

6. Why are different disk partitioning strategies used for read-heavy
and write-heavy workloads?
  - Different strategies are used for Read and Write workloads because their nature is different:
  
  1. Read-Heavy Workloads -
  Tasks that need to retrieve large amount of data quickly.
  Ex - SSDs are ideal because they offer faster read speeds than traditional HDDs.

 Strategy: Large partition, XFS file system, RAID 1 / RAID 10 for fast reading.

 2. Write-Heavy Workloads -

 These tasks frequently modify or add data, often used for speed and data safety.
 Ex - Storing large amount of data or backup in the cloud.


Strategy: Small and separate partitions like /var/log, ext4 journaling file system, RAID 10 and noatime to reduce extra writes.
